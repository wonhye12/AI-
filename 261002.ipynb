```python
def asmd(x,y,op):
    if op=='+':
        print(f'{x} + {y} = {add(x,y)}')
    elif op=='-':
        print(f'{x} - {y} = {sub(x,y)}')
    elif op=='*':  
        print(f'{x} * {y} = {mul(x, y)}')
    elif op=='/':
        print(f'{x} / {y} = {div(x, y):.2f}')
    else:
        print("잘못된 연산자입니다.")


x=int(input('첫 번재 수: '))
y=int(input('두 번재 수: '))
op=input('연산자를 입력하시오: ')
asmd(x,y,op)
```
